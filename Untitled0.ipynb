```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Install required packages
!pip install ta xgboost -q

# ML Libraries
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score

# Technical Indicators - using 'ta' library (more stable)
import ta

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Import for file upload in Colab
from google.colab import files

print("="*80)
print("STOCK PRICE PREDICTION - ML MODEL")
print("="*80)

# ==================== DATA LOADING ====================
print("\n STEP 1: Loading Dataset")
print("-" * 80)

# Option 1: Direct file upload (RECOMMENDED)
print(" Please upload your stock CSV file (e.g., RELIANCE.csv, TCS.csv, INFY.csv)")
print(" You can download Indian stock data from:")
print("   https://www.kaggle.com/datasets/rohanrao/nifty50-stock-market-data\n")

uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print(f"\n Dataset loaded successfully: {filename}")
print(f"   Shape: {df.shape}")
print(f"\nFirst 5 rows:")
print(df.head())

# ==================== DATA PREPROCESSING ====================
print("\n\n🔧 STEP 2: Data Preprocessing ")
print("-" * 80)

# Standardize column names
# === THIS IS THE FIX ===
df.columns = df.columns.str.strip().str.title()
# =======================

# Handle different date column names
date_cols = ['Date', 'Datetime', 'Time', 'Timestamp']
for col in date_cols:
    if col in df.columns:
        df['Date'] = pd.to_datetime(df[col])
        break

# Standardize OHLCV column names
column_mapping = {
    'Open': 'Open', 'High': 'High', 'Low': 'Low',
    'Close': 'Close', 'Volume': 'Volume',
    'Adj Close': 'Adj_Close', 'Close Price': 'Close'
}

for old_name, new_name in column_mapping.items():
    if old_name in df.columns:
        df[new_name] = df[old_name]

# Sort by date
df = df.sort_values('Date')
df.reset_index(drop=True, inplace=True)

print(f" Data preprocessed")
print(f"   Date range: {df['Date'].min()} to {df['Date'].max()}")


# === NEW: EDA SECTION START ===
# ==================== EXPLORATORY DATA ANALYSIS (EDA) ====================
print("\n\n STEP 2.5: Exploratory Data Analysis (EDA)")
print("-" * 80)

# Plot 1: Close Price Over Time
print(" Plotting Close Price and Volume over time...")
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 10), sharex=True,
                               gridspec_kw={'height_ratios': [3, 1]})

# Plot Close Price
ax1.plot(df['Date'], df['Close'], label='Close Price', color='blue')
ax1.set_ylabel('Close Price (₹)', fontweight='bold')
ax1.set_title(f'{filename} - Close Price Over Time', fontweight='bold', fontsize=14)
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot Volume
ax2.bar(df['Date'], df['Volume'], label='Volume', color='gray')
ax2.set_ylabel('Volume', fontweight='bold')
ax2.set_xlabel('Date', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Plot 2: Distribution of Daily Returns
print(" Plotting Distribution of Daily Returns...")
# Calculate daily returns
df['Daily_Return'] = df['Close'].pct_change()

plt.figure(figsize=(10, 6))
sns.histplot(df['Daily_Return'].dropna(), bins=100, kde=True, color='green')

mean_ret = df['Daily_Return'].mean()
std_ret = df['Daily_Return'].std()

plt.axvline(mean_ret, color='red', linestyle='--', label=f'Mean: {mean_ret:.4f}')
plt.axvline(mean_ret + std_ret, color='orange', linestyle=':', label=f'Std Dev: {std_ret:.4f}')
plt.axvline(mean_ret - std_ret, color='orange', linestyle=':')

plt.title('Distribution of Daily Returns', fontweight='bold', fontsize=14)
plt.xlabel('Daily Return', fontweight='bold')
plt.ylabel('Frequency', fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(" EDA plots generated.")
# === NEW: EDA SECTION END ===


# ==================== TECHNICAL INDICATORS ====================
print("\n\n STEP 3: Calculating Technical Indicators")
print("-" * 80)

# Add all ta features at once (more stable approach)
# MACD
macd_indicator = ta.trend.MACD(df['Close'], window_slow=26, window_fast=12, window_sign=9)
df['MACD'] = macd_indicator.macd()
df['MACD_signal'] = macd_indicator.macd_signal()
df['MACD_hist'] = macd_indicator.macd_diff()

# RSI
df['RSI'] = ta.momentum.RSIIndicator(df['Close'], window=14).rsi()

# SMA
df['SMA_20'] = ta.trend.SMAIndicator(df['Close'], window=20).sma_indicator()
df['SMA_50'] = ta.trend.SMAIndicator(df['Close'], window=50).sma_indicator()

# EMA
df['EMA_12'] = ta.trend.EMAIndicator(df['Close'], window=12).ema_indicator()
df['EMA_26'] = ta.trend.EMAIndicator(df['Close'], window=26).ema_indicator()

# Bollinger Bands
bollinger = ta.volatility.BollingerBands(df['Close'], window=20, window_dev=2)
df['BB_upper'] = bollinger.bollinger_hband()
df['BB_middle'] = bollinger.bollinger_mavg()
df['BB_lower'] = bollinger.bollinger_lband()

# ATR
df['ATR'] = ta.volatility.AverageTrueRange(df['High'], df['Low'], df['Close'], window=14).average_true_range()

# OBV
df['OBV'] = ta.volume.OnBalanceVolumeIndicator(df['Close'], df['Volume']).on_balance_volume()

# ADX
df['ADX'] = ta.trend.ADXIndicator(df['High'], df['Low'], df['Close'], window=14).adx()

print(" Technical indicators calculated:")
print("   MACD, RSI, SMA (20, 50), EMA (12, 26), Bollinger Bands, ATR, OBV, ADX")

# ==================== FEATURE ENGINEERING ====================
print("\n\n STEP 4: Feature Engineering")
print("-" * 80)

# Target Variable - Next Day Price Movement
df['Next_Close'] = df['Close'].shift(-1)
df['Target'] = (df['Next_Close'] > df['Close']).astype(int)

# Price and Volume changes
df['Price_Change'] = df['Close'].pct_change()
df['Volume_Change'] = df['Volume'].pct_change()

# Lag features
for lag in [1, 2, 3, 5]:
    df[f'Close_Lag_{lag}'] = df['Close'].shift(lag)
    df[f'Volume_Lag_{lag}'] = df['Volume'].shift(lag)
    df[f'RSI_Lag_{lag}'] = df['RSI'].shift(lag)

# Rolling statistics
df['Close_Rolling_Mean_5'] = df['Close'].rolling(window=5).mean()
df['Close_Rolling_Std_5'] = df['Close'].rolling(window=5).std()
df['Volume_Rolling_Mean_5'] = df['Volume'].rolling(window=5).mean()
df['High_Low_Ratio'] = df['High'] / df['Low']
df['Close_Open_Ratio'] = df['Close'] / df['Open']

# Drop NaN values
df_clean = df.dropna()

print(f" Features engineered")
print(f"   Original samples: {len(df)}")
print(f"   Clean samples: {len(df_clean)}")
print(f"\n Target Distribution:")
print(df_clean['Target'].value_counts())
print(f"\n   Up days: {df_clean['Target'].sum()} ({df_clean['Target'].mean()*100:.2f}%)")
print(f"   Down days: {len(df_clean) - df_clean['Target'].sum()} ({(1-df_clean['Target'].mean())*100:.2f}%)")

# ==================== FEATURE SELECTION ====================
feature_columns = [
    'Open', 'High', 'Low', 'Close', 'Volume',
    'MACD', 'MACD_signal', 'MACD_hist',
    'RSI', 'SMA_20', 'SMA_50', 'EMA_12', 'EMA_26',
    'BB_upper', 'BB_middle', 'BB_lower',
    'ATR', 'OBV', 'ADX',
    'Price_Change', 'Volume_Change',
    'Close_Lag_1', 'Close_Lag_2', 'Close_Lag_3', 'Close_Lag_5',
    'Volume_Lag_1', 'Volume_Lag_2', 'Volume_Lag_3', 'Volume_Lag_5',
    'RSI_Lag_1', 'RSI_Lag_2', 'RSI_Lag_3', 'RSI_Lag_5',
    'Close_Rolling_Mean_5', 'Close_Rolling_Std_5', 'Volume_Rolling_Mean_5',
    'High_Low_Ratio', 'Close_Open_Ratio'
]

X = df_clean[feature_columns]
y = df_clean['Target']


# === NEW: CORRELATION HEATMAP START ===
# ==================== EDA - FEATURE CORRELATION ====================
print("\n\n STEP 4.5: EDA - Feature Correlation")
print("-" * 80)

print(" Generating feature correlation heatmap...")

# Calculate correlation matrix
corr_matrix = X.corr()

plt.figure(figsize=(20, 16))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Feature Correlation Matrix', fontweight='bold', fontsize=16)
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

print(" Correlation heatmap generated.")
# === NEW: CORRELATION HEATMAP END ===


# ==================== TRAIN-TEST SPLIT ====================
print("\n\n STEP 5: Train-Test Split")
print("-" * 80)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False, random_state=42
)

print(f" Data split completed:")
print(f"   Training samples: {len(X_train)} ({len(X_train)/len(X)*100:.1f}%)")
print(f"   Testing samples: {len(X_test)} ({len(X_test)/len(X)*100:.1f}%)")

# Feature Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(" Features scaled using StandardScaler")

# ==================== MODEL TRAINING ====================
print("\n\n" + "="*80)
print(" STEP 6: MODEL TRAINING & EVALUATION")
print("="*80)

results = {}

# 1. Logistic Regression
print("\n1️ Logistic Regression")
print("-" * 40)
lr_model = LogisticRegression(max_iter=1000, random_state=42, C=1.0, solver='lbfgs')
lr_model.fit(X_train_scaled, y_train)
lr_pred = lr_model.predict(X_test_scaled)
lr_acc = accuracy_score(y_test, lr_pred)
results['Logistic Regression'] = {
    'accuracy': lr_acc,
    'precision': precision_score(y_test, lr_pred),
    'recall': recall_score(y_test, lr_pred),
    'f1': f1_score(y_test, lr_pred)
}
print(f"   ✓ Accuracy:  {lr_acc:.4f}")
print(f"   ✓ Precision: {results['Logistic Regression']['precision']:.4f}")
print(f"   ✓ Recall:    {results['Logistic Regression']['recall']:.4f}")
print(f"   ✓ F1-Score:  {results['Logistic Regression']['f1']:.4f}")

# 2. Random Forest
print("\n2️ Random Forest Classifier")
print("-" * 40)
rf_model = RandomForestClassifier(
    n_estimators=200, max_depth=10, min_samples_split=5,
    min_samples_leaf=2, random_state=42, n_jobs=-1
)
rf_model.fit(X_train_scaled, y_train)
rf_pred = rf_model.predict(X_test_scaled)
rf_acc = accuracy_score(y_test, rf_pred)
results['Random Forest'] = {
    'accuracy': rf_acc,
    'precision': precision_score(y_test, rf_pred),
    'recall': recall_score(y_test, rf_pred),
    'f1': f1_score(y_test, rf_pred)
}
print(f"   ✓ Accuracy:  {rf_acc:.4f}")
print(f"   ✓ Precision: {results['Random Forest']['precision']:.4f}")
print(f"   ✓ Recall:    {results['Random Forest']['recall']:.4f}")
print(f"   ✓ F1-Score:  {results['Random Forest']['f1']:.4f}")

# 3. Gradient Boosting
print("\n3️ Gradient Boosting Classifier")
print("-" * 40)
gb_model = GradientBoostingClassifier(
    n_estimators=100, learning_rate=0.1, max_depth=5,
    min_samples_split=5, min_samples_leaf=2, random_state=42
)
gb_model.fit(X_train_scaled, y_train)
gb_pred = gb_model.predict(X_test_scaled)
gb_acc = accuracy_score(y_test, gb_pred)
results['Gradient Boosting'] = {
    'accuracy': gb_acc,
    'precision': precision_score(y_test, gb_pred),
    'recall': recall_score(y_test, gb_pred),
    'f1': f1_score(y_test, gb_pred)
}
print(f"   ✓ Accuracy:  {gb_acc:.4f}")
print(f"   ✓ Precision: {results['Gradient Boosting']['precision']:.4f}")
print(f"   ✓ Recall:    {results['Gradient Boosting']['recall']:.4f}")
print(f"   ✓ F1-Score:  {results['Gradient Boosting']['f1']:.4f}")

# 4. XGBoost
print("\n4️ XGBoost Classifier")
print("-" * 40)
xgb_model = XGBClassifier(
    n_estimators=100, learning_rate=0.1, max_depth=5,
    min_child_weight=3, gamma=0.1, subsample=0.8,
    colsample_bytree=0.8, random_state=42, n_jobs=-1,
    eval_metric='logloss'
)
xgb_model.fit(X_train_scaled, y_train)
xgb_pred = xgb_model.predict(X_test_scaled)
xgb_acc = accuracy_score(y_test, xgb_pred)
results['XGBoost'] = {
    'accuracy': xgb_acc,
    'precision': precision_score(y_test, xgb_pred),
    'recall': recall_score(y_test, xgb_pred),
    'f1': f1_score(y_test, xgb_pred)
}
print(f"   ✓ Accuracy:  {xgb_acc:.4f}")
print(f"   ✓ Precision: {results['XGBoost']['precision']:.4f}")
print(f"   ✓ Recall:    {results['XGBoost']['recall']:.4f}")
print(f"   ✓ F1-Score:  {results['XGBoost']['f1']:.4f}")

# 5. Support Vector Machine
print("\n5️ Support Vector Machine (SVM)")
print("-" * 40)
svm_model = SVC(C=1.0, kernel='rbf', gamma='scale', random_state=42, probability=True) # Added probability=True
svm_model.fit(X_train_scaled, y_train)
svm_pred = svm_model.predict(X_test_scaled)
svm_acc = accuracy_score(y_test, svm_pred)
results['SVM'] = {
    'accuracy': svm_acc,
    'precision': precision_score(y_test, svm_pred),
    'recall': recall_score(y_test, svm_pred),
    'f1': f1_score(y_test, svm_pred)
}
print(f"   ✓ Accuracy:  {svm_acc:.4f}")
print(f"   ✓ Precision: {results['SVM']['precision']:.4f}")
print(f"   ✓ Recall:    {results['SVM']['recall']:.4f}")
print(f"   ✓ F1-Score:  {results['SVM']['f1']:.4f}")

# 6. K-Nearest Neighbors
print("\n6 K-Nearest Neighbors (KNN)")
print("-" * 40)
knn_model = KNeighborsClassifier(n_neighbors=5, weights='distance', metric='euclidean')
knn_model.fit(X_train_scaled, y_train)
knn_pred = knn_model.predict(X_test_scaled)
knn_acc = accuracy_score(y_test, knn_pred)
results['KNN'] = {
    'accuracy': knn_acc,
    'precision': precision_score(y_test, knn_pred),
    'recall': recall_score(y_test, knn_pred),
    'f1': f1_score(y_test, knn_pred)
}
print(f"   ✓ Accuracy:  {knn_acc:.4f}")
print(f"   ✓ Precision: {results['KNN']['precision']:.4f}")
print(f"   ✓ Recall:    {results['KNN']['recall']:.4f}")
print(f"   ✓ F1-Score:  {results['KNN']['f1']:.4f}")

# ==================== RESULTS SUMMARY ====================
print("\n\n" + "="*80)
print(" RESULTS SUMMARY")
print("="*80)

results_df = pd.DataFrame(results).T
results_df = results_df.sort_values('accuracy', ascending=False)

print("\nHere is the performance table for all trained models:")
print(results_df)

best_model_name = results_df.index[0]
best_accuracy = results_df.iloc[0]['accuracy']

print(f"\n\n BEST MODEL: {best_model_name}")
print(f"   Accuracy: {best_accuracy:.4f} ({best_accuracy*100:.2f}%)")

# Select best model predictions
model_predictions = {
    'Logistic Regression': lr_pred,
    'Random Forest': rf_pred,
    'Gradient Boosting': gb_pred,
    'XGBoost': xgb_pred,
    'SVM': svm_pred,
    'KNN': knn_pred
}
best_pred = model_predictions[best_model_name]

# Detailed Classification Report
print("\n\n" + "="*80)
print(f" DETAILED CLASSIFICATION REPORT - {best_model_name}")
print("="*80)
print(classification_report(y_test, best_pred, target_names=['Down (0)', 'Up (1)']))

# Confusion Matrix
print("\n Confusion Matrix:")
cm = confusion_matrix(y_test, best_pred)
print(cm)
print(f"\nTrue Negatives:  {cm[0,0]}")
print(f"False Positives: {cm[0,1]}")
print(f"False Negatives: {cm[1,0]}")
print(f"True Positives:  {cm[1,1]}")

# Feature Importance
model_objects = {
    'Random Forest': rf_model,
    'Gradient Boosting': gb_model,
    'XGBoost': xgb_model
}

if best_model_name in model_objects:
    print("\n\n" + "="*80)
    print(f" TOP 15 FEATURE IMPORTANCES - {best_model_name}")
    print("="*80)

    importances = model_objects[best_model_name].feature_importances_
    feature_importance_df = pd.DataFrame({
        'Feature': feature_columns,
        'Importance': importances
    }).sort_values('Importance', ascending=False)

    print("\n" + feature_importance_df.head(15).to_string(index=False))

# ==================== VISUALIZATION ====================
print("\n\n Generating visualizations...")

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Stock Price Prediction - Model Performance Analysis', fontsize=16, fontweight='bold')

# Plot 1: Model Accuracy Comparison
ax1 = axes[0, 0]
colors = ['#2ecc71' if x == best_accuracy else '#3498db' for x in results_df['accuracy']]
bars = ax1.barh(results_df.index, results_df['accuracy'], color=colors)
ax1.set_xlabel('Accuracy', fontweight='bold')
ax1.set_title('Model Accuracy Comparison', fontweight='bold')
ax1.set_xlim([0, 1])
for i, v in enumerate(results_df['accuracy']):
    ax1.text(v + 0.01, i, f'{v:.4f}', va='center', fontweight='bold')
ax1.grid(axis='x', alpha=0.3)

# Plot 2: Confusion Matrix
ax2 = axes[0, 1]
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax2, cbar_kws={'label': 'Count'})
ax2.set_title(f'Confusion Matrix - {best_model_name}', fontweight='bold')
ax2.set_ylabel('Actual', fontweight='bold')
ax2.set_xlabel('Predicted', fontweight='bold')
ax2.set_xticklabels(['Down', 'Up'])
ax2.set_yticklabels(['Down', 'Up'])

# Plot 3: Metrics Comparison
ax3 = axes[0, 2]
metrics = results_df[['accuracy', 'precision', 'recall', 'f1']].loc[best_model_name]
bars = ax3.bar(metrics.index, metrics.values, color=['#e74c3c', '#f39c12', '#9b59b6', '#1abc9c'])
ax3.set_ylabel('Score', fontweight='bold')
ax3.set_title(f'Performance Metrics - {best_model_name}', fontweight='bold')
ax3.set_ylim([0, 1])
for i, v in enumerate(metrics.values):
    ax3.text(i, v + 0.02, f'{v:.3f}', ha='center', fontweight='bold')
ax3.grid(axis='y', alpha=0.3)

# Plot 4: Feature Importance (if applicable)
ax4 = axes[1, 0]
if best_model_name in model_objects:
    top_features = feature_importance_df.head(10)
    bars = ax4.barh(top_features['Feature'], top_features['Importance'], color='#16a085')
    ax4.set_xlabel('Importance', fontweight='bold')
    ax4.set_title('Top 10 Feature Importances', fontweight='bold')
    ax4.grid(axis='x', alpha=0.3)
else:
    ax4.text(0.5, 0.5, f'Feature importance\nnot available for\n{best_model_name}',
             ha='center', va='center', fontsize=12)
    ax4.axis('off')

# Plot 5: Prediction vs Actual
ax5 = axes[1, 1]
sample_size = min(100, len(y_test))
x_range = range(sample_size)
ax5.plot(x_range, y_test.values[:sample_size], label='Actual',
         marker='o', linestyle='-', markersize=4, linewidth=2, color='#e74c3c')
ax5.plot(x_range, best_pred[:sample_size], label='Predicted',
         marker='s', linestyle='--', markersize=4, linewidth=2, color='#3498db', alpha=0.7)
ax5.set_xlabel('Sample Index', fontweight='bold')
ax5.set_ylabel('Class (0=Down, 1=Up)', fontweight='bold')
ax5.set_title(f'Actual vs Predicted (First {sample_size} samples)', fontweight='bold')
ax5.legend(loc='best')
ax5.grid(True, alpha=0.3)
ax5.set_yticks([0, 1])

# Plot 6: All Models F1-Score Comparison
ax6 = axes[1, 2]
f1_scores = results_df['f1'].sort_values(ascending=True)
colors_f1 = ['#2ecc71' if i == len(f1_scores)-1 else '#95a5a6' for i in range(len(f1_scores))]
bars = ax6.barh(f1_scores.index, f1_scores.values, color=colors_f1)
ax6.set_xlabel('F1-Score', fontweight='bold')
ax6.set_title('Model F1-Score Comparison', fontweight='bold')
ax6.set_xlim([0, 1])
for i, v in enumerate(f1_scores.values):
    ax6.text(v + 0.01, i, f'{v:.3f}', va='center', fontweight='bold')
ax6.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('stock_prediction_results.png', dpi=300, bbox_inches='tight')
print(" Visualization saved as 'stock_prediction_results.png'")
plt.show()

# ==================== NEXT DAY PRICE PREDICTION ====================
print("\n\n" + "="*80)
print(" NEXT DAY PRICE PREDICTION")
print("="*80)

# Get the latest data point for prediction
latest_data = df_clean.iloc[-1:]
latest_close = latest_data['Close'].values[0]
latest_date = latest_data['Date'].values[0]

print(f"\n Latest Trading Day: {pd.to_datetime(latest_date).strftime('%Y-%m-%d')}")
print(f" Latest Close Price: ₹{latest_close:.2f}")

# Prepare features for prediction
latest_features = latest_data[feature_columns]
latest_features_scaled = scaler.transform(latest_features)

# Get predictions from all models
print("\n" + "-" * 80)
print("Model Predictions (1 = Price Up ↗, 0 = Price Down ↘):")
print("-" * 80)

predictions = {}
models_dict = {
    'Logistic Regression': lr_model,
    'Random Forest': rf_model,
    'Gradient Boosting': gb_model,
    'XGBoost': xgb_model,
    'SVM': svm_model,
    'KNN': knn_model
}

for model_name, model in models_dict.items():
    pred = model.predict(latest_features_scaled)[0]
    predictions[model_name] = pred
    direction = "UP ↗" if pred == 1 else "DOWN ↘"
    color = "🟢" if pred == 1 else "🔴"
    print(f"{color} {model_name:20s}: {direction}")

# Ensemble prediction (majority voting)
pred_values = list(predictions.values())
ensemble_prediction = 1 if sum(pred_values) > len(pred_values) / 2 else 0
ensemble_confidence = sum(pred_values) / len(pred_values) * 100

print("\n" + "="*80)
print(" ENSEMBLE PREDICTION (Majority Voting)")
print("="*80)

if ensemble_prediction == 1:
    print(f"\n Prediction: Price will GO UP ↗")
    print(f" Confidence: {ensemble_confidence:.1f}% ({sum(pred_values)}/{len(pred_values)} models agree)")
else:
    print(f"\n  Prediction: Price will GO DOWN ↘")
    print(f" Confidence: {100-ensemble_confidence:.1f}% ({len(pred_values)-sum(pred_values)}/{len(pred_values)} models agree)")

# Best model prediction
best_model_obj = models_dict[best_model_name]
best_pred_direction = best_model_obj.predict(latest_features_scaled)[0]

print("\n" + "="*80)
print(f" BEST MODEL PREDICTION ({best_model_name})")
print("="*80)

if best_pred_direction == 1:
    print(f"\n Prediction: Price will GO UP ↗")
else:
    print(f"\n  Prediction: Price will GO DOWN ↘")

# Calculate estimated price range based on historical volatility
price_changes = df_clean['Close'].pct_change().dropna()
avg_change = price_changes.mean()
std_change = price_changes.std()

if ensemble_prediction == 1:
    expected_change = avg_change + (std_change * 0.5)  # Positive scenario
    lower_estimate = latest_close * (1 + (avg_change - std_change))
    upper_estimate = latest_close * (1 + (avg_change + std_change))
else:
    expected_change = avg_change - (std_change * 0.5)  # Negative scenario
    lower_estimate = latest_close * (1 + (avg_change - std_change))
    upper_estimate = latest_close * (1 + (avg_change + std_change))

predicted_price = latest_close * (1 + expected_change)

print("\n" + "-" * 80)
print(" ESTIMATED NEXT DAY PRICE RANGE")
print("-" * 80)
print(f"\n   Current Price:     ₹{latest_close:.2f}")
print(f"   Predicted Price:   ₹{predicted_price:.2f}")
print(f"   Expected Change:   {expected_change*100:+.2f}%")
print(f"\n    Confidence Range:")
print(f"   Lower Estimate:    ₹{lower_estimate:.2f}")
print(f"   Upper Estimate:    ₹{upper_estimate:.2f}")

# Probability estimation for tree-based models
models_with_proba = ['Random Forest', 'Gradient Boosting', 'XGBoost', 'Logistic Regression', 'SVM']
if best_model_name in models_with_proba:
    try:
        proba = best_model_obj.predict_proba(latest_features_scaled)[0]
        print(f"\n   🎲 Probability Estimates ({best_model_name}):")
        print(f"   Price Down: {proba[0]*100:.1f}%")
        print(f"   Price Up:   {proba[1]*100:.1f}%")
    except Exception as e:
        print(f"\n   Could not get probability for {best_model_name}: {e}")
        pass

print("\n\n" + "="*80)
print("  IMPORTANT DISCLAIMER")
print("="*80)
print("""
  This prediction is based on historical data and technical indicators.
  Stock markets are inherently unpredictable and influenced by many factors:
      • Global economic events
      • Company news and earnings
      • Political developments
      • Market sentiment
      • Unexpected events

  This model should NOT be the sole basis for investment decisions.
  Always consult with financial advisors and do thorough research.
  Past performance does not guarantee future results.
""")

print("\n" + "="*80)
print(" TRAINING COMPLETED SUCCESSFULLY!")
print("="*80)
print(f"\n Summary:")
print(f"   • Dataset: {filename}")
print(f"   • Total samples: {len(df_clean)}")
print(f"   • Features used: {len(feature_columns)}")
print(f"   • Best model: {best_model_name}")
print(f"   • Best accuracy: {best_accuracy:.4f} ({best_accuracy*100:.2f}%)")
print(f"   • Next day prediction: {'UP ↗' if ensemble_prediction == 1 else 'DOWN ↘'}")
print(f"   • Estimated price: ₹{predicted_price:.2f}")
print("\n" + "="*80)
```
